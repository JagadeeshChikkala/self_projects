## PCA finds a new set of dimensions, called Principal Components (PCs), which are orthogonal (perpendicular) to each other and ordered so that the first PC captures the largest possible variance, the second PC captures the next largest, and so on. You can then choose to keep only the top k components to reduce the number of features in your dataset.

Standardization/Scaling: Since PCA is sensitive to the scale of features, you must first standardize your data (mean = 0, variance = 1) so that all variables contribute equally to the analysis.
Covariance Matrix Computation: Calculate the covariance matrix of the scaled data to understand how the variables relate to each other.
Eigenvalues and Eigenvectors: Compute the eigenvectors and eigenvalues from the covariance matrix.Eigenvectors define the direction of the new Principal Components.Eigenvalues indicate the magnitude or amount of variance captured along that eigenvector's direction.Selecting Components: Sort the eigenvectors by their corresponding eigenvalues in descending order. You then select the top $k$ eigenvectors (the Principal Components) that capture a sufficient percentage of the total variance (e.g., 95%).
Projection: Transform the original data onto the new lower-dimensional subspace defined by the selected $k$ Principal Components.
